```python
numeric_columns = [
    "id",
    "size",
    "stargazers_count",
    "forks_count"
]

for column in numeric_columns:
    print(
        column,
        "| Negative Values:",
        (df[column] < 0).sum()
    )
```

```python
date_columns = ["created_at", "updated_at", "pushed_at"]

for column in date_columns:
    converted_dates = pd.to_datetime(
        df[column],
        errors="coerce",
        utc=True
    )

    print(
        column,
        "| Invalid Dates:",
        converted_dates.isna().sum()
    )
```

```python
def validate_languages(value):
    try:
        data = json.loads(value)
        return isinstance(data, dict)
    except (ValueError, TypeError):
        return False

df["languages_valid"] = df["languages"].apply(validate_languages)

print("Valid Language Records:", df["languages_valid"].sum())
print("Invalid Language Records:", (~df["languages_valid"]).sum())
```

```python
quality_report = {
    "Total Records": len(df),
    "Total Columns": len(df.columns) - 1,
    "Missing Values": df.isnull().sum().sum(),
    "Duplicate Rows": df.duplicated().sum(),
    "Duplicate IDs": df["id"].duplicated().sum(),
    "Unique Repository IDs": df["id"].nunique(),
    "Invalid Language JSON": (~df["languages_valid"]).sum()
}

for check, result in quality_report.items():
    print(f"{check}: {result}")
```

```python
assert len(df) > 0, "Dataset is empty!"
assert df["id"].is_unique, "Duplicate repository IDs found!"
assert df.isnull().sum().sum() == 0, "Missing values found!"
assert df["languages_valid"].all(), "Invalid language JSON found!"

print("All basic data quality tests passed!")
```
